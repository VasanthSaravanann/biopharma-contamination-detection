## Title: Real-Time Microbial Contamination Detection in Biopharmaceutical Manufacturing via Multimodal Sensor Fusion, Explainable AI, and Generative Data Augmentation
**Authors:** Vasanth S, Varsha E, Keerthana S, Swathi RB
**Affiliations:** IIT Madras, RMK CET, St Joseph's College of Engineering, LPU

### Abstract
Microbial contamination in biopharmaceutical manufacturing leads to significant batch failures and patient safety risks. Conventional compendial sterility tests require 14 days for confirmation, precluding real-time intervention. This study presents an integrated UV-Vis spectroscopy and generative AI platform trained on 947 real spectra from the MIT dataset (Rajeev J. Ram, Sci Rep 2025). Three unsupervised anomaly detectors — Isolation Forest, deep autoencoder, and One-Class SVM — are trained exclusively on nominal process spectra. A Multimodal Hierarchical Denoising Diffusion Probabilistic Model (MH-DDPM) synthesizes realistic contamination spectra for rare organisms. SHAP explainability maps detection decisions to biologically meaningful wavelengths. The platform targets detection within 30 minutes at concentrations as low as 10 CFU/mL. Real-data results, time-series early-warning curves, and multimodal fusion analysis are presented for the first time.

## 1. Introduction
Biopharmaceutical manufacturing requires stringent quality control. Industry analyses indicate $1-2M USD losses per contamination incident (FDA, 2025). The compendial sterility testing method (USP ⟨71⟩) mandates 14-day incubation, inherently precluding real-time intervention.

## 2. Methods
### 2.1 Dataset
947 UV-Vis spectra (230-600 nm) from Rajeev J. Ram (MIT/SMART), Scientific Reports 2025 (DOI: 10.1038/s41598-024-83114-y). Categories: 497 sterile, 379 contaminated, 71 timepoint. Organisms: E. coli, S. aureus, B. subtilis, C. albicans, C. sporogenes, P. aeruginosa, C. acnes. CFU range: 0-1000 CFU/mL.

### 2.2 Anomaly Detection
Three unsupervised models trained on clean-only data: Isolation Forest (n_estimators=200), One-Class SVM (RBF kernel, γ=0.001), Deep Autoencoder (128→64→16 latent). Weighted ensemble with cross-validated weights.

### 2.3 MH-DDPM
Denoising Diffusion Probabilistic Model with 50 timesteps, trained on 379 contaminated spectra. Generates synthetic spectra conditioned on organism type.

### 2.4 Explainability
SHAP TreeExplainer on Random Forest classifier to identify wavelength importance. Biological mapping to DNA, protein, pH, biomass absorption bands.
